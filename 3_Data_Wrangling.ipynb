```python
import pandas as pd
import numpy as np



df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')

print("--- 1. TỔNG QUAN DỮ LIỆU GỐC ---")
print(f"Số lượng bản ghi: {df.shape[0]}, Số cột đặc trưng: {df.shape[1]}")


print("\n--- 2. KIỂM TRA GIÁ TRỊ THIẾU (MISSING VALUES) ---")
print(df.isnull().sum())


mean_payload = df['PayloadMass'].mean()
df['PayloadMass'].fillna(mean_payload, inplace=True)


print("\n--- 3. THỐNG KÊ KẾT QUẢ HẠ CÁNH (OUTCOME) ---")
landing_outcomes = df['Outcome'].value_counts()
print(landing_outcomes)


bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
print("\nDanh sách các trạng thái hạ cánh thất bại (bad_outcomes):")
print(bad_outcomes)


landing_class = []
for outcome in df['Outcome']:
    if outcome in bad_outcomes:
        landing_class.append(0)
    else:
        landing_class.append(1)

df['Class'] = landing_class

print("\n--- 4. TỶ LỆ HẠ CÁNH THÀNH CÔNG TRUNG BÌNH ---")
print(f"Tỷ lệ thành công (Class = 1): {df['Class'].mean():.2%}")



features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 
               'GridFins', 'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]


features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'])


features_one_hot = features_one_hot.astype('float64')


df.to_csv('dataset_part_2.csv', index=False)
features_one_hot.to_csv('dataset_part_3.csv', index=False)

print("\n-----------------------------------------------------")
print("DATA WRANGLING HOÀN TẤT THÀNH CÔNG!")
print(f"- Kích thước Dataset đã gán nhãn Class (dataset_part_2.csv): {df.shape}")
print(f"- Kích thước Ma trận đặc trưng One-Hot (dataset_part_3.csv): {features_one_hot.shape}")
```
