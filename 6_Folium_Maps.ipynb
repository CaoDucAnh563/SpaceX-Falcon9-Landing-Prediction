```python
import folium
import pandas as pd
from folium.plugins import MarkerCluster
from math import sin, cos, sqrt, atan2, radians


def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0 
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c


spacex_df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv')


site_map = folium.Map(location=[28.562302, -80.577356], zoom_start=5)


marker_cluster = MarkerCluster().add_to(site_map)

for index, row in spacex_df.iterrows():
    
    marker_color = 'green' if row['class'] == 1 else 'red'
    folium.Marker(
        location=[row['Lat'], row['Long']],
        icon=folium.Icon(color='white', icon_color=marker_color),
        popup=f"Site: {row['Launch Site']} | Class: {row['class']}"
    ).add_to(marker_cluster)


launch_site_lat, launch_site_lon = 28.56319, -80.57683
coastline_lat, coastline_lon = 28.56367, -80.57163 

distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)


distance_marker = folium.Marker(
    [coastline_lat, coastline_lon],
    icon=folium.DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html=f'<div style="font-size: 12pt; color: #d9534f;"><b>{distance_coastline:.2f} KM</b></div>'
    )
).add_to(site_map)


lines = folium.PolyLine(locations=[[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]], weight=2, color='blue')
site_map.add_child(lines)


site_map.save('spacex_folium_map.html')
print("Khởi tạo bản đồ Folium hoàn tất! Đã xuất file HTML 'spacex_folium_map.html'.")
site_map
```
