```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv')


sns.set_theme(style="whitegrid")


plt.figure(figsize=(12, 6))
sns.scatterplot(x="FlightNumber", y="LaunchSite", hue="Class", data=df, palette={0: "red", 1: "green"}, s=100)
plt.title("Biểu đồ 1: Số hiệu chuyến bay và Bãi phóng (phân loại theo Kết quả hạ cánh)", fontsize=14)
plt.xlabel("Flight Number", fontsize=12)
plt.ylabel("Launch Site", fontsize=12)
plt.show()


plt.figure(figsize=(12, 6))
sns.scatterplot(x="PayloadMass", y="LaunchSite", hue="Class", data=df, palette={0: "red", 1: "green"}, s=100)
plt.title("Biểu đồ 2: Khối lượng tải trọng vs Bãi phóng", fontsize=14)
plt.xlabel("Payload Mass (kg)", fontsize=12)
plt.ylabel("Launch Site", fontsize=12)
plt.show()


plt.figure(figsize=(10, 6))
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index()
sns.barplot(x='Orbit', y='Class', data=orbit_success, palette="viridis")
plt.title("Biểu đồ 3: Tỷ lệ hạ cánh thành công trung bình theo dạng Quỹ đạo (Orbit)", fontsize=14)
plt.xlabel("Orbit Type", fontsize=12)
plt.ylabel("Success Rate (0 - 1.0)", fontsize=12)
plt.ylim(0, 1.1)
plt.show()


plt.figure(figsize=(12, 6))
sns.scatterplot(x="FlightNumber", y="Orbit", hue="Class", data=df, palette={0: "red", 1: "green"}, s=100)
plt.title("Biểu đồ 4: Số hiệu chuyến bay vs Dạng quỹ đạo", fontsize=14)
plt.xlabel("Flight Number", fontsize=12)
plt.ylabel("Orbit", fontsize=12)
plt.show()


plt.figure(figsize=(12, 6))
sns.scatterplot(x="PayloadMass", y="Orbit", hue="Class", data=df, palette={0: "red", 1: "green"}, s=100)
plt.title("Biểu đồ 5: Khối lượng tải trọng vs Dạng quỹ đạo", fontsize=14)
plt.xlabel("Payload Mass (kg)", fontsize=12)
plt.ylabel("Orbit", fontsize=12)
plt.show()


df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly_success = df.groupby('Year')['Class'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(x='Year', y='Class', data=yearly_success, marker='o', color='b', linewidth=2.5)
plt.title("Biểu đồ 6: Xu hướng tỷ lệ hạ cánh thành công theo năm (2010 - 2020)", fontsize=14)
plt.xlabel("Năm (Year)", fontsize=12)
plt.ylabel("Tỷ lệ thành công trung bình", fontsize=12)
plt.grid(True)
plt.show()

print("EDA Visualization hoàn tất! Đã xuất 6 biểu đồ phục vụ Slide nộp bài.")
```
