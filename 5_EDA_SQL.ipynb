```python
import sqlite3
import pandas as pd


spacex_csv_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv'
df_sql = pd.read_csv(spacex_csv_url)


df_sql.columns = df_sql.columns.str.replace(' ', '_')


conn = sqlite3.connect("spacex.db")
df_sql.to_sql("SPACEXTBL", conn, if_exists='replace', index=False)

def run_query(query_text, title=""):
    print(f"\n==================== {title} ====================")
    print(f"SQL: {query_text}\n")
    result = pd.read_sql_query(query_text, conn)
    print(result)
    return result


run_query("SELECT DISTINCT Launch_Site FROM SPACEXTBL;", "1. Danh sách bãi phóng duy nhất")


run_query("SELECT * FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5;", "2. 5 Chuyến phóng tại CCA")


run_query("SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_Mass FROM SPACEXTBL WHERE Customer = 'NASA (CRS)';", "3. Tổng tải trọng NASA (CRS)")


run_query("SELECT AVG(PAYLOAD_MASS__KG_) AS Avg_Payload_Mass FROM SPACEXTBL WHERE Booster_Version LIKE 'F9 v1.1%';", "4. Tải trọng trung bình F9 v1.1")


run_query("SELECT MIN(Date) AS First_Successful_Ground_Landing FROM SPACEXTBL WHERE Mission_Outcome LIKE 'Success%';", "5. Chuyến thành công đầu tiên")


run_query("SELECT Booster_Version FROM SPACEXTBL WHERE Landing_Outcome = 'Success (drone ship)' AND PAYLOAD_MASS__KG_ BETWEEN 4000 AND 6000;", "6. Booster tải trọng 4000-6000kg thành công drone ship")


run_query("SELECT Mission_Outcome, COUNT(*) AS Count FROM SPACEXTBL GROUP BY Mission_Outcome;", "7. Thống kê kết quả nhiệm vụ")


run_query("SELECT Booster_Version, PAYLOAD_MASS__KG_ FROM SPACEXTBL WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL);", "8. Booster chở tải trọng tối đa")


run_query("SELECT substr(Date, 6, 2) AS Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTBL WHERE substr(Date, 1, 4)='2015' AND Landing_Outcome = 'Failure (drone ship)';", "9. Thất bại Drone Ship năm 2015")


run_query("""
SELECT Landing_Outcome, COUNT(*) AS Outcome_Count 
FROM SPACEXTBL 
WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' 
GROUP BY Landing_Outcome 
ORDER BY Outcome_Count DESC;
""", "10. Xếp hạng kết quả hạ cánh (2010-2017)")

conn.close()
print("\n--- EDA với SQL Hoàn Tất Thành Công! ---")
```
